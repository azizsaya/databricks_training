```python
%sql
CREATE OR REPLACE VIEW dev.aziz.agg_heartrate AS (
  SELECT mrn, name, MEAN(heartrate) avg_heartrate, DATE_TRUNC("DD", time) date
  FROM dev.aziz.heartrate_device
  GROUP BY mrn, name, DATE_TRUNC("DD", time)
);
SELECT * FROM dev.aziz.agg_heartrate
```

```python
%sql
CREATE OR REPLACE FUNCTION dev.aziz.datamask(x STRING)
  RETURNS STRING
  RETURN CONCAT(REPEAT("*", LENGTH(x) - 2), RIGHT(x, 2)
); 
```

```python
%sql
CREATE OR REPLACE VIEW dev.aziz.agg_heartrate AS (
  SELECT 
  CASE WHEN
    is_account_group_member('account users') THEN dev.aziz.datamask(mrn)
    ELSE mrn
  END AS mrn, 
  name, 
  MEAN(heartrate) avg_heartrate, 
  DATE_TRUNC("DD", time) date
  FROM dev.aziz.heartrate_device
  GROUP BY mrn, name, DATE_TRUNC("DD", time)
);
```

```python
%sql
SELECT * FROM dev.aziz.agg_heartrate
```
